## Explore The Data Pattern

### Check Survival Rate by Categories

```python
## Survival Rate by Sex
sns.barplot(x='sex',y='survived',data=titanic)
plt.show()

## Survival by ticket class
sns.barplot(x='pclass',y='survived',data=titanic)
plt.show()
```

```python
## Check The Survival Rate by Age and Pclass

sns.boxplot(x='pclass',y='age',hue='survived',data=titanic)
plt.show()
```

```python
## Check age distribution

sns.histplot(titanic['age'],bins=30,kde=True)
plt.show()
```

### Check Fare Distribution

```python
sns.histplot(titanic['fare'],bins=30,kde=True)
plt.title("Distribution of Ticket Price")
plt.xlabel("Fare")
plt.ylabel("No of Passenger")
```
