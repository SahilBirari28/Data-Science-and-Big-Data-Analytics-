```python
text = "Tokenization is the process of splitting sentences into words help to study studying studies better and best good computer"
```

## Part 1 :  Document Preprocessing

### Tokenization

```python
from nltk.tokenize import word_tokenize
tokenize_words = word_tokenize(text)
tokenize_words
```

### Remove Stopwords

```python
from nltk.corpus import stopwords
```

```python
sw = set(stopwords.words("english"))
print(sw)
```

```python
filtered_words = []
for word in tokenize_words:
    if word.lower() not in sw:
        filtered_words.append(word)

filtered_words
```

### Regular Expression

```python
import re
text1 = "How to remove Stopword,punctuation from nltk library in pyhton ? "

text2 = re.sub('[^a-zA-z]',' ',text1)
```

```python
text2
```

### POS Tagging

```python
from nltk import pos_tag
print(pos_tag(filtered_words),"\n")
```

### Stemming

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
for word in filtered_words:
    print(ps.stem(word),end = ' ')
```

### Lemmatization

```python
from nltk.stem import WordNetLemmatizer
le = WordNetLemmatizer()
for w in filtered_words:
    print(w, " : ",le.lemmatize(w))
```

## Part 2 : Text Representation

### Term Frequency(TF)

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer


clean_doc = ["Data is very is big data very very"]

vect = CountVectorizer()
X = vect.fit_transform(clean_doc)

words = vect.get_feature_names_out()
counts = X.toarray()[0]

df1 = pd.DataFrame({
    "Word": words,
    "TF-IDF": counts
})
df1
```

### Inverse Document Frequency

```python

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(clean_doc)

words = tfidf.get_feature_names_out()
values = X_tfidf.toarray()[0]

# Create DataFrame
df = pd.DataFrame({
    "Word": words,
    "TF-IDF": values
})

df
```

```python
documentA="Jupyter is the largest planet"
documentB="Mars is the 4th planet from the sun"

## Bag of words 
bagOfWordsA= documentA.split(' ')
bagOfWordsB= documentB.split(' ')

## Create collection of unique search
uniqueWords = set(bagOfWordsA).union(bagOfWordsB)
uniqueWords
```

```python
## Unique documents from the document A
numsOfWords1 = dict.fromkeys(uniqueWords,0)
for word in bagOfWordsA:
    numsOfWords1[word]+=1
numsOfWords1
```

```python
## Unique documents from the document B
numsOfWords2 = dict.fromkeys(uniqueWords,0)
for word in bagOfWordsB:
    numsOfWords2[word]+=1
numsOfWords2
```

```python
def computeTF(wordDict,bagOfWords):
    TFDict ={}
    bagcount = len(bagOfWords)
    for word,count in wordDict.items():
        TFDict[word] = count/float(bagcount)
    return TFDict
```

```python
TFA = computeTF(numsOfWords1, bagOfWordsA)
```

```python
TFA
```

```python
TFB= computeTF(numsOfWords2, bagOfWordsB)
TFB
```
