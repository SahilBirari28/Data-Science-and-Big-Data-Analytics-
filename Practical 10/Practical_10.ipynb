```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# For better visuals
sns.set(style="whitegrid")
```

```python
iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Map numeric labels to actual species names
df['species'] = df['species'].map({
    0: 'Setosa',
    1: 'Versicolor',
    2: 'Virginica'
})

df.head()
```

```python
print("Shape of dataset:", df.shape)
print("\nColumn Names:\n", df.columns)
print("\nData Types:\n", df.dtypes)
print("\nSummary Statistics:\n", df.describe())
```

```python
for col in df.columns:
    if df[col].dtype == 'object':
        print(f"{col}: Categorical (Nominal)")
    else:
        print(f"{col}: Numeric (Continuous)")
```

```python
df.hist(figsize=(10, 8), bins=20, edgecolor='black')
plt.suptitle("Histograms of Iris Features", fontsize=18)
plt.show()
```

```python
plt.figure(figsize=(12, 8))

for i, col in enumerate(df.columns[:-1], 1):
    plt.subplot(2, 2, i)
    sns.boxplot(x='species', y=col, data=df)
    plt.title(col)

plt.tight_layout()
plt.show()
```

### Observations:

1. All input features are numeric and continuous.
2. Species is a categorical (nominal) variable.
3. Petal length and petal width show clear separation between species.
4. Sepal features show overlap and are less useful for classification.
5. Sepal width has some outliers.
6. Histograms of petal features show multi-modal distribution.
