```python
titanic = sns.load_dataset("titanic")
```

```python
## Survival Rate by Sex
sns.barplot(x='sex',y='survived',data=titanic)
plt.show()

```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='sex', y='age', hue='survived', data=titanic)
```

```python
# Create age bins
bins = [0, 12, 18, 35, 60, 100]
labels = ['Child', 'Teen', 'Young Adult', 'Adult', 'Senior']

titanic['age_group'] = pd.cut(titanic['age'], bins=bins, labels=labels)

plt.figure(figsize=(8,6))

sns.barplot(x='age_group', y='survived', data=titanic)

plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")

plt.show()
```

##   Observations

You can write something like this in your assignment:

###  Key Insights from the Box Plot

- **Females had higher survival rates**  
  - Most female passengers survived compared to males.  
  - This supports the *“women and children first”* evacuation policy.

- **Male survival rate was lower**  
  - A large portion of male passengers did not survive.  
  - Even younger males had lower survival compared to females.

- **Age distribution differences**  
  - Survivors include a wide age range, but:  
    - Younger passengers (children) had better survival chances.  
    - Older passengers had lower survival probability.

- **Children were prioritized**  
  - Lower median age in survivors indicates that children were more likely to survive.

- **Outliers present**  
  - Some very young or very old passengers survived/did not survive, visible as outliers in the boxplot.

```python
## Check The Survival Rate by Age and Pclass

sns.boxplot(x='pclass',y='age',hue='survived',data=titanic)
plt.show()
```

## Observations: Survival Rate by Age and Passenger Class (Pclass)

### Key Insights from the Box Plot

- **Higher class passengers had better survival rates**  
  - Passengers in 1st class (pclass = 1) show a higher proportion of survivors compared to 2nd and 3rd class.  
  - This indicates that wealthier passengers were given priority or had better access to lifeboats.

- **3rd class passengers had the lowest survival rate**  
  - A large number of non-survivors belong to 3rd class.  
  - This suggests limited access to safety measures for lower-class passengers.

- **Age distribution varies across classes**  
  - 1st class passengers tend to be older on average.  
  - 3rd class includes more younger passengers, including children.

- **Children had better survival chances across classes**  
  - Lower age values (children) appear more frequently among survivors in all classes.  
  - This again supports the “children first” rescue priority.

- **Median age differences**  
  - Survivors in higher classes often have a slightly higher median age than survivors in lower classes.  
  - Indicates that class played a stronger role than age alone.

- **Outliers are present**  
  - Some older passengers survived and some younger passengers did not, visible as outliers in the plot.  
  - Survival was influenced by multiple factors, not just age or class.
